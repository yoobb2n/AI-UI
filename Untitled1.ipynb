## 코디 추천 AI 모델을 위한 가상 데이터 생성 계획

옷 데이터가 없는 상황에서 코디 추천 AI 모델을 개발하기 위해 다음과 같은 단계를 진행할 수 있습니다.

1.  **가상 옷 데이터 생성**: 다양한 옷 아이템과 속성을 포함하는 가상 데이터를 만듭니다.
2.  **사용자 스케줄 및 날씨 데이터 정의**: 스케줄과 날씨를 모델이 활용할 수 있는 형태로 수치화하는 방법을 제안합니다.
3.  **추천 모델 아이디어 구상**: 가상 데이터와 스케줄/날씨 정보를 활용하여 코디를 추천하는 모델의 기본적인 로직을 설명합니다.

### 1. 가상 옷 데이터 생성

먼저, `pandas` 라이브러리를 사용하여 코디 추천에 필요한 기본적인 옷 정보를 담은 가상 데이터프레임을 생성해 보겠습니다. 여기에는 옷의 종류, 색상, 계절 적합성, 스타일 등의 속성이 포함될 것입니다.

````python
import pandas as pd
import numpy as np

# 가상 옷 데이터 생성
np.random.seed(42) # 재현성을 위해 시드 설정

num_items = 100 # 생성할 옷 아이템 수

clothes_data = {
    'item_id': range(1, num_items + 1),
    'category': np.random.choice(['상의', '하의', '아우터', '신발', '액세서리'], num_items, p=[0.3, 0.25, 0.2, 0.15, 0.1]),
    'type': np.random.choice(['티셔츠', '셔츠', '니트', '바지', '스커트', '재킷', '코트', '운동화', '구두', '모자', '가방'], num_items),
    'color': np.random.choice(['블랙', '화이트', '그레이', '네이비', '베이지', '카키', '핑크', '블루', '브라운'], num_items),
    'material': np.random.choice(['면', '울', '폴리에스터', '데님', '가죽', '린넨'], num_items),
    'season_suitability': np.random.choice(['봄/가을', '여름', '겨울', '사계절'], num_items, p=[0.3, 0.2, 0.3, 0.2]),
    'style': np.random.choice(['캐주얼', '포멀', '스트릿', '미니멀', '빈티지', '스포츠'], num_items, p=[0.3, 0.2, 0.15, 0.15, 0.1, 0.1]),
    'warmth_level': np.random.randint(1, 6, num_items), # 1(얇음) ~ 5(매우 두꺼움)
    'water_resistance': np.random.choice([0, 1], num_items, p=[0.8, 0.2]) # 0: 없음, 1: 있음
}

clothes_df = pd.DataFrame(clothes_data)

# 'type' 컬럼을 'category'에 맞게 조정 (예: 상의 카테고리인데 바지 type이 나오지 않도록)
def adjust_type_by_category(row):
    if row['category'] == '상의':
        return np.random.choice(['티셔츠', '셔츠', '니트'])
    elif row['category'] == '하의':
        return np.random.choice(['바지', '스커트', '반바지'])
    elif row['category'] == '아우터':
        return np.random.choice(['재킷', '코트', '패딩', '가디건'])
    elif row['category'] == '신발':
        return np.random.choice(['운동화', '구두', '샌들', '부츠'])
    elif row['category'] == '액세서리':
        return np.random.choice(['모자', '가방', '목도리', '장갑'])
    return row['type'] # 기본값

clothes_df['type'] = clothes_df.apply(adjust_type_by_category, axis=1)

print("가상 옷 데이터프레임의 첫 5행:")
display(clothes_df.head())
print(f"\n총 {len(clothes_df)}개의 옷 아이템 생성.")
````

### 2. 사용자 스케줄 및 날씨 데이터 정의

사용자의 스케줄과 날씨는 코디 추천에 중요한 영향을 미칩니다. 이들을 모델이 이해할 수 있는 수치형 데이터로 변환하는 방법을 고려해 봅시다.

#### 2.1. 사용자 스케줄 데이터

스케줄은 `활동 유형`, `포멀리티`, `활동 시간` 등으로 구성될 수 있습니다. 예를 들어, `'회의'`는 `'포멀'`하고 `'오전'`에 진행될 수 있으며, `'피크닉'`은 `'캐주얼'`하고 `'오후'`에 진행될 수 있습니다.

#### 2.2. 날씨 데이터

날씨는 `온도`, `강수량`, `바람`, `습도` 등 여러 요소로 구성됩니다. 모델에 입력하기 위해 이러한 요소들을 수치화해야 합니다. 특히, 온도는 옷의 `warmth_level`과 직접적인 관련이 있으며, 강수량은 `water_resistance`에 영향을 줄 수 있습니다.

이러한 요소들을 가상으로 생성하여 추천 모델의 입력으로 활용해 보겠습니다.

````python
import pandas as pd
import numpy as np

# 가상 사용자 스케줄 및 날씨 데이터 생성
np.random.seed(42) # 재현성을 위해 시드 설정

num_days = 30 # 30일치 데이터 생성

# 사용자 스케줄 데이터
schedule_data = {
    'date': pd.to_datetime(pd.date_range(start='2023-10-01', periods=num_days)),
    'event_type': np.random.choice(['업무 (사무실)', '업무 (외근)', '친목 모임', '운동/레저', '휴식/집콕', '데이트'], num_days, p=[0.25, 0.15, 0.2, 0.15, 0.15, 0.1]),
    'formality_level': np.random.randint(1, 6, num_days), # 1(매우 캐주얼) ~ 5(매우 포멀)
    'activity_intensity': np.random.randint(1, 4, num_days) # 1(낮음) ~ 3(높음)
}
schedule_df = pd.DataFrame(schedule_data)

# 가상 날씨 데이터 (서울 기준 예시)
weather_data = {
    'date': pd.to_datetime(pd.date_range(start='2023-10-01', periods=num_days)),
    'temperature_celsius': np.random.randint(5, 25, num_days), # 5~24도
    'precipitation_mm': np.random.choice([0, 0, 0, 0, 0, 0, 1, 5, 10], num_days, p=[0.7, 0.1, 0.05, 0.05, 0.02, 0.02, 0.02, 0.02, 0.02]), # 0mm, 1mm, 5mm, 10mm
    'wind_speed_mps': np.random.uniform(0.5, 8.0, num_days).round(1), # 0.5 ~ 8.0 m/s
    'humidity_percent': np.random.randint(40, 95, num_days) # 40% ~ 95%
}
weather_df = pd.DataFrame(weather_data)

print("\n가상 사용자 스케줄 데이터프레임의 첫 5행:")
display(schedule_df.head())

print("\n가상 날씨 데이터프레임의 첫 5행:")
display(weather_df.head())
````

### 3. 추천 모델 아이디어 구상: 가중치 기반 코디 추천 시스템

이제 가상 옷 데이터, 사용자 스케줄, 날씨 데이터를 모두 가지고 있으므로, 이들을 활용하여 코디를 추천하는 기본적인 모델을 구상해 볼 수 있습니다. 제안하는 모델은 `가중치 기반 접근 방식`으로, 각 요소가 추천에 미치는 영향력을 조절하여 사용자 맞춤형 코디를 제공합니다.

#### 3.1. 코디 추천 로직 개요

1.  **사용자 입력 및 맥락 파악**: 특정 날짜의 사용자 스케줄(활동 유형, 포멀리티, 활동 강도)과 해당 날짜의 날씨(온도, 강수량, 바람, 습도)를 입력으로 받습니다.
2.  **옷 아이템 필터링 및 점수화**: 전체 `clothes_df`에서 각 옷 아이템이 현재 맥락(스케줄, 날씨)에 얼마나 적합한지 점수를 매깁니다.
    *   **날씨 관련 가중치**: `temperature_celsius`에 따라 `warmth_level`이 적절한 옷에 높은 점수를 부여합니다. `precipitation_mm`이 높으면 `water_resistance`가 있는 옷에 가산점을 줍니다.
    *   **스케줄 관련 가중치**: `formality_level`에 따라 `style`이 적절한 옷에 높은 점수를 부여합니다. `activity_intensity`에 따라 편안함이나 활동성이 좋은 옷에 가산점을 줍니다.
3.  **최종 코디 조합**: 점수가 높은 옷 아이템들을 조합하여 하나의 코디를 구성합니다. 일반적으로 상의, 하의, 아우터, 신발, 액세서리 등 카테고리별로 점수가 높은 아이템을 선택하여 추천합니다.
